```python
# if we want to see how many cars having diff engine type we use cross tab
pd.crosstab(car_sales['Brand'],car_sales['Engine Type'])

#in this we will see there are 214 audi cars having diesel engine and 53 audi cars having gas engine and 167 audi cars having peteol engine
# jiske number chahiye woh pehle likhnge aur jiske type mein chahiye like kitne hai uss type ke tph usko baad mein likhnge
```

```python
# average price of each brand using groupby
car_sales.groupby('Brand')['Price'].mean()
# hr brand yaani bmw ek group mein audi ek group mein toyota ek group mein phir un sbki average price nikali
```

```python
# min price of each brand using
car_sales.groupby('Brand')['Price'].min()
```

```python
# hr brand ka max min mean sab ek sath dekhna hai
car_sales.groupby('Brand')['Price'].agg(['mean','max','min','sum'])
```
